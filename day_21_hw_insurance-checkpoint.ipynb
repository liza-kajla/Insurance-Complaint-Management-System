```python
# here, we will inroduce CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# call and assign
cv = CountVectorizer()
```

```python
# view
cv
```

---

```python
# model building
```

```python
# model building
from sklearn.naive_bayes import MultinomialNB
```

```python
# call assign
mnb = MultinomialNB()
```

```python
# view
mnb
```

```python
# let us define X and Y
df[:2]
```

```python
# Use only text column
X = df["Complaint_Text"]

# Convert ALL target labels to string
y = df["Complaint_Category"].astype(str)
```

```python
X[:2]
```

```python
# view
y[:10]
```

```python
y = y.replace({
    "Policy Cancellation": 0,
    "Billing Dispute": 1,
    "Coverage Denial": 2,
    "Customer Service Issue": 3,
    "Fraud Suspicion": 4,
    "Claim Delay": 5
})
```

```python
# view
y[:5]
```

```python
# model selection
from sklearn.model_selection import train_test_split
```

```python
# apply
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

---

```python
pool_train = cv.fit_transform(X_train)
pool_test = cv.transform(X_test)
```

---

```python
# train the model
mnb.fit(pool_train, y_train)
```

```python
mnb
```

```python
# training score
mnb.score(pool_train, y_train)
```

```python
# testing score
mnb.score(pool_test, y_test)
```

```python
# prediction
y_pred = mnb.predict(pool_test)
```

```python
y_pred
```

---

```python
# model evaluation 
from sklearn.metrics import classification_report, confusion_matrix,accuracy_score
```

```python
# classification report
print(classification_report(y_test,y_pred))
```

```python
# accuracy score
accuracy_score(y_test,y_pred)
```

```python
# confusion matrix

cm = confusion_matrix(y_test, y_pred)
cm
```

```python
# graph for confusion matrix
plt.figure(figsize=(16,4))
sns.heatmap(cm,annot=True, cmap ="Blues")
plt.title("Confusion Matrix")
plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.show()
```

```python
# for streamlit, we need to save the vectorizer and the model
import joblib

# save the trained model
joblib.dump(mnb, "mnb.pkl")

# save the CountVectorizer (must be the SAME one used to transform X_train)
joblib.dump(cv, "cv.pkl")
```
